```python
#Install libraries
%matplotlib inline
from preamble import *
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

#Read csv file
df = pd.read_csv('Cred_app_reviews.csv')
df.head()

print(df.shape)
```

```python
#Display all columns
df.columns
```

```python
#Drop any rows that contain at least one empty cell
df = df.dropna(how='any')
```

```python
#Print the shape of the resulting dataset
print(df.shape)
```

```python
df.head()
```

```python
import pandas as pd
import nltk
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

#These downloads are only necessary the first time
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

#Access the text and score columns from the dataset
text_data = df["content"]
score_data = df["score"]

#Pre-processing
tokens = []
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words("english"))

for text in text_data:
    tokenized_text = word_tokenize(text)
    filtered_tokens = [token.lower() for token in tokenized_text if token.lower() not in stop_words]
    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]
    tokens.append(" ".join(lemmatized_tokens))

#Feature extraction
vectorizer = TfidfVectorizer()
features = vectorizer.fit_transform(tokens)

#Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(features, score_data, test_size=0.2, random_state=42)

#Train a linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

#Make a prediction using the test set
y_pred = model.predict(X_test)

#Evaluate the model using mean squared error
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)
```
